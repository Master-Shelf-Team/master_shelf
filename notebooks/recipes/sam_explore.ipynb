### Imports

```python
import pandas as pd
import numpy as np
```

### Chargement du CSV

```python
data = pd.read_csv("../../raw_data/recipes_ingredients.csv")
data.set_index(keys="id", inplace=True)
data.head()
```

```python
import ast

errors = []

def safe_literal_eval(value):
    try:
     return ast.literal_eval(value)
    except (ValueError, SyntaxError) as e:
        errors.append((value, str(e)))
    return value

data["ingredients"] = data["ingredients"].apply(safe_literal_eval)
```

```python
mot = "chicken"
data_filtre = data[data['ingredients'].apply(lambda x: any(mot in ing.lower() for ing in x))]
data_filtre
```
