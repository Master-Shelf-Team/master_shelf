## Les colonnes ingredients, ingredients_raw, steps, tags sont des strings qui contiennent des listes.
## --> il faut les transformer en listes de strings qui pourront êtres extraites et exploitées

(le split ne fonctionne pas ici)

```python
import ast
```

```python
errors=[]

def return_list(text):
    try :
        return ast.literal_eval(text)
    except (ValueError, SyntaxError) as e:
        errors.append((text, str(e)))
        return text
```

```python
return_list(df.loc[0,'ingredients'])
```

```python
df[['ingredients', 'ingredients_raw', 'steps', 'tags']] = df[['ingredients', 'ingredients_raw', 'steps', 'tags']] .map(return_list)
```
